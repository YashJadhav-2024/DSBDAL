```python
def age_group(age):
    if(age<=18):
        return 'Prodigy'
    elif(age>=18 and age<=25):
        return"Young player"
    elif(age>25 and age<=35):
        return"Experienced player"
    else:
        return'Legend player'
        
df['player_age']=df['Age'].apply(age_group)
```

```python
categorical_var = "player_age"
stats = (
df.groupby(categorical_var)["Salary"]
.agg(['mean', 'median', 'min', 'max', 'std'])
)
stats
#this shows that an experienced player that whose age is between 25 and 3
#has the higher salary owing to their experience
#also legends have higher salary and young players due to their experience
```

```python
categorical_var = "Position"
stats = (
df.groupby(categorical_var)["Age"]
.agg(['mean', 'median', 'min', 'max', 'std'])
)
stats
#the center position players have the highest mean age because
#they need experience to play in that position
#shooting gaurds are generally runners so they need to be young
#therefore they have the lowest mean age
```
