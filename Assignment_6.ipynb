```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
df=pd.read_csv('iris.csv')
```

```python
df.head()
```

```python
#list down all festures
df.info()
```

```python
#list down various types of petals
df['variety'].unique()
```

```python
#Visualize each feature to illustrate  the feature distribution

#feature distribution of sepal.length

sns.histplot(data=df,x='sepal.length',hue='variety',kde=True)

#Virginica has larget variety of sepal length than others
```

```python
#feature distribution of sepal.width
sns.histplot(data=df,x='sepal.width',hue='variety',kde=True)

#Setosa has larger variety of sepal width than others while versicolor has the smallest
```

```python
#feature distribution of petal.length
sns.histplot(data=df,x='petal.length',hue='variety',kde=True)

#Virginica has the largest petal length while Setosa has smallest
```

```python
#feature distribution of petal.width
sns.histplot(data=df,x='petal.width',hue='variety')

#Virginica has largest petal width among others
```

```python
#Create a boxplot for each feature in dataset
```

```python
#boxpot for sepal length
sns.boxplot(data=df,y='sepal.length',x='variety')

#Outliers detected in virginica sepal length
```

```python
#boxplot for sepal width
sns.boxplot(data=df,y='sepal.width',x='variety')

#outliers detected in Setosa and Virginica and median of Setosa sepal width is greater than others
```

```python
#boxplot for petal length
sns.boxplot(data=df,y='petal.length',x='variety')
#Too many outlier sin the petal length of Setosa compared to others also setosa median petal; length is significantly lower.
```

```python
#boxplot for petal width
sns.boxplot(data=df,y='petal.width',x='variety')

#Too many outlier sin the petal width of Setosa compared to others also setosa median petal; length is significantly lower.
```
