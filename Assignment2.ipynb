```python
from scipy import stats
```

```python
z = np.abs(stats.zscore(tf['math score']))
print(z)
```

```python
threshold = 2.5
sample_outliers = np.where(z > threshold)
sample_outliers
```

```python
sorted_rscore= sorted(tf['reading score'])
sorted_rscore
```

```python
q1 = np.percentile(sorted_rscore, 25)
q3 = np.percentile(sorted_rscore, 75)
print(q1,q3)
```

```python
IQR = q3-q1
lwr_bound = q1-(1.5*IQR)
upr_bound = q3+(1.5*IQR)
print(lwr_bound, upr_bound)
```

```python
r_outliers = []
for i in sorted_rscore:
     if (i<lwr_bound or i>upr_bound):
       r_outliers.append(i)
print(r_outliers)
```

```python
new_tf=tf
for i in sample_outliers:
  new_tf.drop(i,inplace=True)
new_tf 
```
